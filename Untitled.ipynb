```python
def mostBluePixelinArray(the_array):#assume RGBA
    """
    Return the index of the "bluest" pixel in an RGBA array, if any blue pixel exists.

    The input is expected to be an array-like object of shape (N, 4), where each row
    represents a pixel in RGBA order. The alpha channel is ignored. For each pixel,
    a simple blueness score is computed as:

        blue / (red + green + blue + 0.01)

    The function tracks the pixel with the highest blueness score and returns its index,
    but only if at least one pixel meets the heuristic for being blue:

        red fraction < 0.45, green fraction < 0.45, and blue fraction >= 0.55

    If no pixel satisfies this blue-pixel condition, the function returns None.

    Args:
        the_array: A NumPy-like array of shape (N, 4) containing RGBA pixel values.

    Returns:
        int or None: The index of the bluest pixel if at least one blue pixel is found,
        otherwise None.
    """
    the_array = the_array[:, 0:3]
    length = len(the_array)
    most_blue_i = 0
    most_blueness = 0.0
    any_blue = False
    for i in range(395, 663):
        tot_rgb = np.sum(the_array[i]) + 0.01
        blueness_i = the_array[i][2]/tot_rgb
        if blueness_i >= most_blueness:
            most_blue_i = i
            most_blueness = blueness_i
        rgb = the_array[i]/tot_rgb
        if(rgb[0] < 0.45 and rgb[1] < 0.45 and rgb[2] >= 0.55):
            any_blue = True
    #print(the_array[most_blue_i])
    if any_blue:
        return most_blue_i
    else:
        return None
    
def mostGreenPixelinArray(the_array):#assume RGBA
    the_array = the_array[:, 0:3]
    length = len(the_array)
    most_blue_i = 0
    most_blueness = 0.0
    any_blue = False
    for i in range(length):
        tot_rgb = np.sum(the_array[i]) + 0.01
        blueness_i = the_array[i][1]/tot_rgb
        if blueness_i >= most_blueness:
            most_blue_i = i
            most_blueness = blueness_i
        rgb = the_array[i]/tot_rgb
        if(rgb[0] < 0.45 and rgb[1] < 0.45 and rgb[2] >= 0.55):
            any_blue = True
    #print(the_array[most_blue_i])
    if any_blue:
        return most_blue_i
    else:
        return None
```

```python
"""
Convert the accumulated purple-point coordinates into a filtered time-series array.

This code:
- Converts `purple_points_over_time` from a list of per-frame coordinate rows into a
  NumPy array.
- Transposes the array so each coordinate component is treated as its own time series.
- Applies frequency-domain low-pass filtering to every odd-indexed row, which
  corresponds to the y-coordinates in the `(x, y)` interleaved structure.
- Leaves even-indexed rows unchanged, which correspond to x-coordinates.
- Uses an FFT-based filter to remove high-frequency noise from the y-coordinate
  trajectories over time.
- Transposes the result back to the original frame-major layout.

Filtering details:
- For each y-coordinate time series:
  - Compute the real FFT (`np.fft.rfft`).
  - Compute the corresponding frequency bins (`np.fft.rfftfreq`).
  - Zero out coefficients above a cutoff set to 40% of the available frequency range.
  - Reconstruct the smoothed signal with the inverse real FFT (`np.fft.irfft`).

Result:
- `filtered_purple` contains the smoothed purple-point coordinates over time.
- `purple_points_over_time` is restored to its original orientation after processing.
"""

purple_points_over_time = np.array(purple_points_over_time)
purple_points_over_time = purple_points_over_time.T
filtered_purple = np.zeros(purple_points_over_time.shape)
for i in range(2*num_rays):
    if (i%2 == 0):
        filtered_purple[i] = purple_points_over_time[i]
    else:
        n = len(purple_points_over_time[i])
        fft_coeffs = np.fft.rfft(purple_points_over_time[i])
        frequencies = np.fft.rfftfreq(n)
        cutoff = np.min(frequencies) + 0.25*(np.max(frequencies) - np.min(frequencies))
        fft_coeffs[frequencies > cutoff] = 0
        filtered_purple[i] = np.fft.irfft(fft_coeffs, n=n)
filtered_purple = filtered_purple.T
purple_points_over_time = purple_points_over_time.T
```

```python
def draw_filtered_points_on_video(path, points, radius=5, color=(255, 0, 127), thickness=2):
    """
    points: (N_frames, 2*N_points) array — each row is [x1, y1, x2, y2, ...]
    """
    """
    Draw a sequence of tracked points onto a video frame-by-frame and overwrite the video.

    This function opens a video at `path`, iterates through its frames, and overlays
    the corresponding point coordinates from `points` onto each frame. The `points`
    array is expected to contain one row per frame, with coordinates stored in the
    interleaved format:

        [x1, y1, x2, y2, ...]

    For each frame, the coordinates are reshaped into `(N_points, 2)` and drawn as
    circles using OpenCV. Frames are written to a temporary video file, which then
    replaces the original video at the end.

    Args:
        path (str): Path to the input video file to annotate.
        points (np.ndarray): Array of shape `(N_frames, 2 * N_points)` containing
            interleaved x/y coordinates for each frame.
        radius (int, optional): Radius of the circles drawn for each point.
            Defaults to 5.
        color (tuple, optional): BGR color of the drawn circles. Defaults to
            `(255, 0, 127)`.
        thickness (int, optional): Circle outline thickness. Defaults to 2.
            Use `-1` to draw filled circles.

    Side Effects:
        - Reads the video at `path`.
        - Writes a temporary annotated video file.
        - Replaces the original video file with the annotated version.

    Notes:
        - Frames without corresponding point data are written unmodified.
        - NaN x-coordinates are skipped.
        - The function assumes `points` is aligned with the video frame order.
    """
    cap = cv2.VideoCapture(path)
    fps = cap.get(cv2.CAP_PROP_FPS)
    w, h = int(cap.get(3)), int(cap.get(4))
    fourcc = cv2.VideoWriter_fourcc(*"mp4v")
    tmp = path + ".tmp.mp4"
    out = cv2.VideoWriter(tmp, fourcc, fps, (w, h))
    i = 0
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        if i < len(points):
            coords = points[i].reshape(-1, 2)
            for x, y in coords:
                if not np.isnan(x):
                    cv2.circle(frame, (int(x), int(y)), radius, color, thickness)
        out.write(frame)
        i += 1
    cap.release()
    out.release()
    import os; os.replace(tmp, path)

unfiltered_video_path = '/mnt/c/Users/rohan/Desktop/GreenUponGreen-DrCarlette-2026-07-17/videos/0532_CLAHE_with_pink_points_with_orange_and_purple_points.mp4'
draw_filtered_points_on_video(unfiltered_video_path, filtered_purple)
```

```python
import pandas as pd
from pathlib import Path
import glob
import sys

project_path = r'/mnt/c/Users/rohan/Desktop/ComputerVisionPipelineTests/Test5/Test5-DrCarlette-2026-08-04/labeled-data/0802_vid/CollectedData_DrCarlette.csv'
def fix_h5(csv_path):
    """
    Convert a DeepLabCut-style multi-index CSV file into an HDF5 file.

    This function reads a CSV file with a three-row header and three index columns
    (as commonly produced by pose-labeling workflows), then saves the data to an
    HDF5 file with the same base filename.

    The CSV is read using:
    - a 3-level column header: `header=[0, 1, 2]`
    - a 3-level row index: `index_col=[0, 1, 2]`

    The output file is written beside the input file with the `.h5` extension and
    stored under the HDF5 key `'df_with_missing'`.

    Args:
        csv_path (str or Path): Path to the input CSV file.

    Side Effects:
        - Reads the CSV from disk.
        - Writes a new `.h5` file to disk, overwriting any existing file with the
          same name.
        - Prints the path of the saved HDF5 file.

    Example:
        >>> fix_h5("CollectedData_DrCarlette.csv")
        Saved: CollectedData_DrCarlette.h5
    """
    p = Path(csv_path)
    df = pd.read_csv(p, header=[0,1,2], index_col=[0,1,2])
    out = p.with_suffix('.h5')
    df.to_hdf(out, key='df_with_missing', mode='w')
    print(f"Saved: {out}")

fix_h5(project_path)
```

```python
"""
Annotate a folder of PNG images using point coordinates stored in a DeepLabCut HDF5 file.

This script:
- Loads the most recent `.h5` file from `h5_dir`.
- Reads x/y coordinate columns from the pose-estimation results, ignoring likelihood.
- Loads all `.png` images from `image_dir` in sorted order.
- For each image/frame:
  - Reads the corresponding image with OpenCV.
  - Draws every tracked point as a filled green circle, regardless of confidence.
  - Saves the annotated image to an `annotated` subfolder.

Behavior:
- Assumes the HDF5 file and image sequence are aligned by frame index.
- Stops processing if there are more images than frames in the HDF5 file.
- Skips images that cannot be read.
- Creates the output directory if it does not already exist.

Side effects:
- Reads an HDF5 file and PNG images from disk.
- Writes annotated PNG images to:
  `image_dir/annotated`
- Prints progress messages for each saved image.
"""

import pandas as pd
import cv2
import numpy as np
import glob
import os

image_dir = r'/mnt/c/Users/rohan/Desktop/ComputerVisionPipelineTests/Test5/Test5-DrCarlette-2026-08-04/labeled-data/0802_vid'
h5_dir = r'/mnt/c/Users/rohan/Desktop/ComputerVisionPipelineTests/Test5/Test5-DrCarlette-2026-08-04/labeled-data/0802_vid'

# Load H5
h5_files = glob.glob(os.path.join(h5_dir, '*.h5'))
df = pd.read_hdf(h5_files[-1])

# Get x/y columns only (ignore likelihood)
x_cols = [col for col in df.columns if col[2] == 'x']
y_cols = [col for col in df.columns if col[2] == 'y']

# Load and sort PNG images
image_paths = sorted(glob.glob(os.path.join(image_dir, '*.png')))

output_dir = os.path.join(image_dir, 'annotated')
os.makedirs(output_dir, exist_ok=True)

for frame_idx, image_path in enumerate(image_paths):
    if frame_idx >= len(df):
        break

    frame = cv2.imread(image_path)
    if frame is None:
        print(f"Could not read image: {image_path}")
        continue

    h, w = frame.shape[:2]

    # Plot every point regardless of confidence
    green_x_pts_list = []
    green_y_pts_list = []
    for x_col, y_col in zip(x_cols, y_cols):
        x = int(df.iloc[frame_idx][x_col])
        y = int(df.iloc[frame_idx][y_col])
        if 0 <= x < w and 0 <= y < h:
            cv2.circle(frame, (x, y), 8, (0, 125, 255), -1)
            green_x_pts_list.append(x)
            green_y_pts_list.append(y)


    out_path = os.path.join(output_dir, os.path.basename(image_path))
    cv2.imwrite(out_path, frame)
    print(f"Saved: {out_path}")

print(f"Done! Annotated images saved to: {output_dir}")
```

```python
import os
import shutil
import glob
import pandas as pd
import numpy as np
from PIL import Image

# ---- Configuration ----
PROJECT_PATH = r"/mnt/c/Users/rohan/Downloads/GreenUponGreen-DrCarlette-2026-07-17/GreenUponGreen-DrCarlette-2026-07-17"
LABELED_DATA_FOLDER = "0532_CLAHE_with_pink_points"
LABELED_DATA_DIR = os.path.join(PROJECT_PATH, "labeled-data", LABELED_DATA_FOLDER)
CSV_PATH = os.path.join(LABELED_DATA_DIR, "CollectedData_DrCarlette.csv")
H5_PATH = os.path.join(LABELED_DATA_DIR, "CollectedData_DrCarlette.h5")

TRANSLATE_PIXELS = 50  # pixels to shift left/right


# ---- Helper functions ----
def flip_image(img_path, flip_type, output_path):
    with Image.open(img_path) as img:
        if flip_type == "horizontal":
            flipped = img.transpose(Image.FLIP_TOP_BOTTOM)
        elif flip_type == "vertical":
            flipped = img.transpose(Image.FLIP_LEFT_RIGHT)
        elif flip_type == "both":
            flipped = img.transpose(Image.ROTATE_180)
        flipped.save(output_path)


def translate_image_wrap(img_path, dx, output_path):
    """Shift image horizontally by dx pixels, wrapping around the edges."""
    with Image.open(img_path) as img:
        w, h = img.size
        arr = np.array(img)
        shift = dx % w
        translated = np.roll(arr, shift, axis=1)
        Image.fromarray(translated).save(output_path)


def flip_keypoints(row_values, flip_type, img_width, img_height):
    new_values = row_values.copy().astype(float)
    num_keypoints = len(new_values) // 2
    for i in range(num_keypoints):
        x_idx = i * 2
        y_idx = i * 2 + 1
        x, y = new_values[x_idx], new_values[y_idx]
        if x == 0 and y == 0:
            continue
        if flip_type == "horizontal":
            new_values[y_idx] = img_height - y
        elif flip_type == "vertical":
            new_values[x_idx] = img_width - x
        elif flip_type == "both":
            new_values[x_idx] = img_width - x
            new_values[y_idx] = img_height - y
    return new_values


def translate_keypoints(row_values, dx, img_width):
    """Shift x-coordinates by dx, wrapping around image width."""
    new_values = row_values.copy().astype(float)
    num_keypoints = len(new_values) // 2
    for i in range(num_keypoints):
        x_idx = i * 2
        y_idx = i * 2 + 1
        x, y = new_values[x_idx], new_values[y_idx]
        if x == 0 and y == 0:
            continue
        new_values[x_idx] = (x + dx) % img_width
    return new_values


# ---- Main ----

# Read original data from CSV (3 header rows, 3 index columns)
df = pd.read_csv(CSV_PATH, header=[0, 1, 2], index_col=[0, 1, 2])
df.index.names = [None, None, None]
df = df.astype(float)
print(f"Original data: {df.shape[0]} rows, {df.shape[1] // 2} keypoints")

# Collect original image names from the index
original_images = [idx[2] for idx in df.index]

new_rows = []
images_created = 0

FLIP_TYPES = {
    "_flipH":  "horizontal",
    "_flipV":  "vertical",
    "_flipHV": "both",
}

for img_name in original_images:
    img_path = os.path.join(LABELED_DATA_DIR, img_name)
    if not os.path.exists(img_path):
        print(f"  WARNING: {img_path} not found, skipping")
        continue

    with Image.open(img_path) as img:
        img_width, img_height = img.size

    idx = ("labeled-data", LABELED_DATA_FOLDER, img_name)
    original_row = df.loc[idx].values
    base_name = os.path.splitext(img_name)[0]
    ext = os.path.splitext(img_name)[1]

    # --- Flips ---
    for suffix, flip_type in FLIP_TYPES.items():
        new_img_name = f"{base_name}{suffix}{ext}"
        new_img_path = os.path.join(LABELED_DATA_DIR, new_img_name)
        flip_image(img_path, flip_type, new_img_path)
        new_values = flip_keypoints(original_row, flip_type, img_width, img_height)
        new_idx = ("labeled-data", LABELED_DATA_FOLDER, new_img_name)
        new_rows.append((new_idx, new_values))
        images_created += 1

    # --- Translations (left and right) ---
    for dx, suffix in [(-TRANSLATE_PIXELS, "_shiftL"), (TRANSLATE_PIXELS, "_shiftR")]:
        new_img_name = f"{base_name}{suffix}{ext}"
        new_img_path = os.path.join(LABELED_DATA_DIR, new_img_name)
        translate_image_wrap(img_path, dx, new_img_path)
        new_values = translate_keypoints(original_row, dx, img_width)
        new_idx = ("labeled-data", LABELED_DATA_FOLDER, new_img_name)
        new_rows.append((new_idx, new_values))
        images_created += 1

print(f"Created {images_created} augmented images")

# Build augmented dataframe
new_indices = [r[0] for r in new_rows]
new_data = np.array([r[1] for r in new_rows])

new_df = pd.DataFrame(
    data=new_data,
    index=pd.MultiIndex.from_tuples(new_indices, names=df.index.names),
    columns=df.columns,
)

augmented_df = pd.concat([df, new_df])
augmented_df.sort_index(inplace=True)

# Backup original CSV
csv_backup = CSV_PATH + ".backup"
if not os.path.exists(csv_backup):
    shutil.copy2(CSV_PATH, csv_backup)
    print("Backup saved as .backup file")

# Save CSV and H5
augmented_df.to_csv(CSV_PATH)
augmented_df.to_hdf(H5_PATH, key="df_with_missing", mode="w")

print(f"\nDone!")
print(f"  Original frames: {len(df)}")
print(f"  New augmented frames: {images_created}")
print(f"  Total labeled frames: {len(augmented_df)}")
```

```python
import os
import shutil
import pandas as pd
import numpy as np
from PIL import Image

# ---- Configuration ----
PROJECT_PATH = r"/mnt/c/Users/rohan/Downloads/GreenUponGreen-DrCarlette-2026-07-17/GreenUponGreen-DrCarlette-2026-07-17"
LABELED_DATA_FOLDER = "0532_CLAHE_with_pink_points"
LABELED_DATA_DIR = os.path.join(PROJECT_PATH, "labeled-data", LABELED_DATA_FOLDER)
CSV_PATH = os.path.join(LABELED_DATA_DIR, "CollectedData_DrCarlette.csv")
H5_PATH = os.path.join(LABELED_DATA_DIR, "CollectedData_DrCarlette.h5")

SHIFT_UP_PIXELS = 50
SUFFIXES_TO_SKIP = ["_flipH", "_flipHV", "_flipV", "_shiftL", "_shiftR", "_shiftU"]


def is_original_image(img_name):
    base = os.path.splitext(img_name)[0]
    return not any(base.endswith(s) for s in SUFFIXES_TO_SKIP)


def shift_image_up(img_path, dy, output_path):
    arr = np.array(Image.open(img_path))
    h = arr.shape[0]
    shifted = np.empty_like(arr)
    shifted[:h - dy] = arr[dy:]          # shift content up
    shifted[h - dy:] = arr[h - dy:]      # fill bottom with original bottom rows
    Image.fromarray(shifted).save(output_path)


def shift_keypoints_up(row_values, dy, img_height):
    new_values = row_values.copy().astype(float)
    num_keypoints = len(new_values) // 2
    for i in range(num_keypoints):
        x_idx = i * 2
        y_idx = i * 2 + 1
        x, y = new_values[x_idx], new_values[y_idx]
        if x == 0 and y == 0:
            continue
        new_values[y_idx] = y - dy
    return new_values


# ---- Main ----
df = pd.read_csv(CSV_PATH, header=[0, 1, 2], index_col=[0, 1, 2])
df.index.names = [None, None, None]
df = df.astype(float)

original_indices = [idx for idx in df.index if is_original_image(idx[2])]
print(f"Found {len(original_indices)} original images (skipping augmented)")

new_rows = []
images_created = 0

for idx in original_indices:
    img_name = idx[2]
    img_path = os.path.join(LABELED_DATA_DIR, img_name)
    if not os.path.exists(img_path):
        print(f"  WARNING: {img_path} not found, skipping")
        continue

    with Image.open(img_path) as img:
        img_width, img_height = img.size

    base_name = os.path.splitext(img_name)[0]
    ext = os.path.splitext(img_name)[1]

    new_img_name = f"{base_name}_shiftU{ext}"
    new_img_path = os.path.join(LABELED_DATA_DIR, new_img_name)

    shift_image_up(img_path, SHIFT_UP_PIXELS, new_img_path)
    new_values = shift_keypoints_up(df.loc[idx].values, SHIFT_UP_PIXELS, img_height)
    new_idx = ("labeled-data", LABELED_DATA_FOLDER, new_img_name)
    new_rows.append((new_idx, new_values))
    images_created += 1

print(f"Created {images_created} shifted images")

new_indices = [r[0] for r in new_rows]
new_data = np.array([r[1] for r in new_rows])

new_df = pd.DataFrame(
    data=new_data,
    index=pd.MultiIndex.from_tuples(new_indices, names=df.index.names),
    columns=df.columns,
)

augmented_df = pd.concat([df, new_df])
augmented_df.sort_index(inplace=True)

# Save
augmented_df.to_csv(CSV_PATH)
augmented_df.to_hdf(H5_PATH, key="df_with_missing", mode="w")

print(f"\nDone!")
print(f"  Previous total: {len(df)}")
print(f"  New shifted frames: {images_created}")
print(f"  New total: {len(augmented_df)}")
print(f"  Training frames (0.99 split): {int(len(augmented_df) * 0.99)}")
```

```python
"""
LightningPose B-spline + FFT low-pass filter pipeline.

Paste this entire cell into a Jupyter notebook.
It will:
  1. Read the LightningPose predictions CSV from the 16-06-30 folder.
  2. Open the labeled video 0532_CLAHE_with_pink_points_labeled.mp4.
  3. For each frame, sort keypoints by x, filter by y-range, fit a quadratic
     baseline + weighted B-spline, and interpolate num_rays purple points.
  4. After all frames are processed, apply an FFT low-pass filter to the
     y-coordinates of the purple points over time.
  5. Write a *copy* of the video with the filtered purple points overlaid.

>>> Adjust `predictions_csv`, `input_video_path`, and `output_video_path`
>>> to match your local paths.
"""

import pandas as pd
import cv2
import numpy as np
import time
import os
import warnings
from numpy.lib.stride_tricks import sliding_window_view
from scipy.interpolate import splrep, splev

warnings.filterwarnings("ignore")

# ─────────────────────────────────────────────
# PATHS — edit these to match your machine
# ─────────────────────────────────────────────
base_dir = r"/mnt/c/Users/rohan/Downloads/16-0"

predictions_csv = os.path.join(
    base_dir, "16-06-30", "video_preds",
    "0532_CLAHE_with_pink_points.csv"
)
input_video_path = os.path.join(
    base_dir, "16-06-30", "video_preds", "labeled_videos",
    "0532_CLAHE_with_pink_points_labeled.mp4"
)
output_video_path = os.path.join(
    base_dir, "16-06-30", "video_preds", "labeled_videos",
    "0532_CLAHE_with_pink_points_labeled_with_purple.mp4"
)

# ─────────────────────────────────────────────
# PARAMETERS
# ─────────────────────────────────────────────
num_rays = 200          # number of interpolated purple points per frame
y_min = 386             # keep only keypoints with y in [y_min, y_max)
y_max = 664
lowpass_cutoff_frac = 0.25  # fraction of frequency range to keep

# ─────────────────────────────────────────────
# 1. LOAD PREDICTIONS CSV
# ─────────────────────────────────────────────
df = pd.read_csv(predictions_csv, header=[0, 1, 2])

# The first column is the frame index; drop it for keypoint access
scorer = df.columns[1][0]  # e.g. "heatmap_mhcrnn_tracker"

# Collect (x_col, y_col) pairs for every bodypart
bodypart_names = []
for col in df.columns[1:]:
    if col[2] == "x":
        bodypart_names.append(col[1])

x_cols = [(scorer, bp, "x") for bp in bodypart_names]
y_cols = [(scorer, bp, "y") for bp in bodypart_names]

print(f"Loaded {len(df)} frames, {len(bodypart_names)} bodyparts")

# ─────────────────────────────────────────────
# 2. OPEN VIDEO
# ─────────────────────────────────────────────
cap = cv2.VideoCapture(input_video_path)
fps = cap.get(cv2.CAP_PROP_FPS)
w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
print(f"Video: {w}x{h} @ {fps} fps, {total_frames} frames")

# ─────────────────────────────────────────────
# 3. PER-FRAME: spline fitting → purple points
# ─────────────────────────────────────────────
purple_points_over_time = []

loop_start = time.perf_counter()
frame_idx = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret or frame_idx >= len(df):
        break

    if frame_idx % 50 == 0:
        print(f"  spline fit: {frame_idx} / {total_frames}")

    # Extract x, y for this frame and sort by x
    xs_raw = np.array([float(df.iloc[frame_idx][xc]) for xc in x_cols])
    ys_raw = np.array([float(df.iloc[frame_idx][yc]) for yc in y_cols])

    # Sort by x
    sort_idx = np.argsort(xs_raw)
    xs_raw = xs_raw[sort_idx]
    ys_raw = ys_raw[sort_idx]

    # Filter by y range and x within frame
    mask = (xs_raw >= 0) & (xs_raw < w) & (ys_raw >= y_min) & (ys_raw < y_max)
    true_x = xs_raw[mask]
    true_y = ys_raw[mask]

    if len(true_x) < 10:
        # Not enough points — emit NaNs for this frame
        purple_row = [np.nan] * (2 * num_rays)
        purple_points_over_time.append(purple_row)
        frame_idx += 1
        continue

    # --- Quadratic baseline ---
    quad_coeffs = np.polyfit(true_x, true_y, 2)
    baseline = np.polyval(quad_coeffs, true_x)
    y_for_spline = true_y - baseline

    # --- Arc-length parameterisation ---
    t_list = np.sqrt(np.diff(baseline) ** 2 + np.diff(true_x) ** 2)
    t_list = np.insert(t_list, 0, 0)
    t_list = np.cumsum(t_list)

    # --- Weights (emphasise body, de-weight tail & head extremes) ---
    spline_point = 0.3 * np.min(true_x) + 0.7 * np.max(true_x)
    body_spline_point = (np.min(true_x) + spline_point) / 2.0
    weights = np.ones(len(true_x))
    weights[true_x < body_spline_point] = 5.0
    weights[(true_x >= body_spline_point) & (true_x <= spline_point)] = 10.0
    weights[true_x > spline_point] = 0.25
    weights[:min(10, len(weights))] = 0.01

    # --- Uniqueness mask (strictly increasing x for splrep) ---
    unique_mask = np.diff(true_x, prepend=-np.inf) > 0

    # --- Optional in-range mask (skip leftmost dark-border region) ---
    head_row_idx = int(true_y[np.argmin(true_x)])
    head_row_idx = np.clip(head_row_idx, 0, h - 1)
    # We don't have `frame` pixel access here per the original code's heuristic,
    # so we skip the first_tan_i filter.  If you need it, uncomment and adapt:
    # that_row = frame[head_row_idx, :, :]
    # first_tan_i = 0
    # for i in range(len(that_row)):
    #     if np.min(that_row[i]) < 200:
    #         first_tan_i = i
    #         break
    # in_range_mask = true_x > (first_tan_i + 113)
    # combined_mask = in_range_mask & unique_mask
    combined_mask = unique_mask

    t_filt = t_list[combined_mask]
    y_filt = y_for_spline[combined_mask]
    w_filt = weights[combined_mask]
    x_filt = true_x[combined_mask]

    if len(t_filt) < 4:
        purple_row = [np.nan] * (2 * num_rays)
        purple_points_over_time.append(purple_row)
        frame_idx += 1
        continue

    # --- Fit B-spline ---
    try:
        tck_y = splrep(t_filt, y_filt, w=w_filt, s=0.05 * len(x_filt))
    except Exception:
        purple_row = [np.nan] * (2 * num_rays)
        purple_points_over_time.append(purple_row)
        frame_idx += 1
        continue


    # --- Evaluate on a smooth grid ---
    x_smooth = np.linspace(np.min(x_filt), np.max(x_filt), num_rays)
    baseline_smooth = np.polyval(quad_coeffs, x_smooth)

    t_smooth = np.sqrt(np.diff(baseline_smooth) ** 2 + np.diff(x_smooth) ** 2)
    t_smooth = np.insert(t_smooth, 0, 0)
    t_smooth = np.cumsum(t_smooth)

    y_smooth = splev(t_smooth, tck_y)

    # Clip to data range
    y_smooth = np.clip(y_smooth, np.min(y_filt), np.max(y_filt))

    
    # --- Sliding-window average ---
    ws = 6
    if len(y_smooth) > ws:
        windows = sliding_window_view(y_smooth, window_shape=ws)
        y_smooth = np.hstack((windows.mean(axis=-1), np.zeros(ws - 1)))
    
        
    y_smooth = baseline_smooth + y_smooth

    # Store interleaved [x1, y1, x2, y2, ...]
    purple_row = []
    for i in range(num_rays):
        purple_row.append(float(x_smooth[i]))
        purple_row.append(float(y_smooth[i]))
    purple_points_over_time.append(purple_row)

    frame_idx += 1

cap.release()
print(f"Spline fitting done ({frame_idx} frames in {time.perf_counter()-loop_start:.1f}s)")

# ─────────────────────────────────────────────
# 4. FFT LOW-PASS FILTER (across time)
# ─────────────────────────────────────────────
purple_points_over_time = np.array(purple_points_over_time, dtype=float)

# Transpose → each row is one coordinate component across all frames
purple_T = purple_points_over_time.T
filtered_purple = np.zeros_like(purple_T)

for i in range(2 * num_rays):
    if i % 2 == 0:
        # x-coordinates: pass through unchanged
        filtered_purple[i] = purple_T[i]
    else:
        # y-coordinates: FFT low-pass filter
        signal = purple_T[i]
        # Handle NaNs: interpolate before FFT, then mask back
        nan_mask = np.isnan(signal)
        if nan_mask.all():
            filtered_purple[i] = signal
            continue
        # Simple linear interpolation over NaN gaps
        if nan_mask.any():
            good = np.where(~nan_mask)[0]
            signal = np.interp(np.arange(len(signal)), good, signal[good])

        n = len(signal)
        fft_coeffs = np.fft.rfft(signal)
        frequencies = np.fft.rfftfreq(n)
        cutoff = np.min(frequencies) + lowpass_cutoff_frac * (
            np.max(frequencies) - np.min(frequencies)
        )
        fft_coeffs[frequencies > cutoff] = 0
        filtered = np.fft.irfft(fft_coeffs, n=n)

        # Restore NaNs so those frames won't be drawn
        filtered[nan_mask] = np.nan
        filtered_purple[i] = filtered

filtered_purple = filtered_purple.T  # back to (N_frames, 2*num_rays)
print("Low-pass filtering done")

# ─────────────────────────────────────────────
# 5. DRAW FILTERED PURPLE POINTS ONTO A COPY
# ─────────────────────────────────────────────
def draw_filtered_points_on_video(
    src_path, dst_path, points,
    radius=5, color=(255, 0, 127), thickness=2
):
    """
    Read `src_path`, overlay `points` on each frame, write to `dst_path`.
    points: (N_frames, 2*N_pts) — interleaved [x1,y1,x2,y2,...].
    """
    cap = cv2.VideoCapture(src_path)
    fps = cap.get(cv2.CAP_PROP_FPS)
    w_v = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    h_v = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fourcc = cv2.VideoWriter_fourcc(*"mp4v")
    out = cv2.VideoWriter(dst_path, fourcc, fps, (w_v, h_v))

    i = 0
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        if i % 50 == 0:
            print(f"  drawing: {i}")
        if i < len(points):
            coords = points[i].reshape(-1, 2)
            for x, y in coords:
                if not np.isnan(x) and not np.isnan(y):
                    cv2.circle(frame, (int(x), int(y)), radius, color, thickness)
        out.write(frame)
        i += 1

    cap.release()
    out.release()
    print(f"Saved annotated video to: {dst_path}")

draw_filtered_points_on_video(input_video_path, output_video_path, filtered_purple)
print("All done!")
```

```python
#with custom function
"""
LightningPose custom-function fitting pipeline.

Paste this entire cell into a Jupyter notebook.
It will:
  1. Read the LightningPose predictions CSV from the 16-06-30 folder.
  2. Open the labeled video 0532_CLAHE_with_pink_points_labeled.mp4.
  3. For each frame, sort keypoints by x, filter by y-range, fit
     custom_function (cubic + exponentially-decayed sinusoids) via curve_fit,
     and generate num_rays smooth purple points.
  4. Write a *copy* of the video with the purple points overlaid.

>>> Adjust `predictions_csv`, `input_video_path`, and `output_video_path`
>>> to match your local paths.
"""

import pandas as pd
import cv2
import numpy as np
import time
import os
import warnings
from scipy.optimize import curve_fit

warnings.filterwarnings("ignore")

# ─────────────────────────────────────────────
# CUSTOM FUNCTION (cubic + exp-decayed sinusoids)
# ─────────────────────────────────────────────
def custom_function(x, a3, a2, a1, a0, a_exp, h_exp, b, w1, p1, w2, p2, a_sin):
    sum_of_sines = a_exp * np.sin(w1 * (x - p1)) + a_sin * np.sin(w2 * (x - p2))
    return (a3 * np.power(x, 3) + a2 * np.power(x, 2) + a1 * x + a0
            + np.exp(-1 * h_exp * (x - 0 * b)) * sum_of_sines)

# ─────────────────────────────────────────────
# PATHS — edit these to match your machine
# ─────────────────────────────────────────────
base_dir = r"/mnt/c/Users/rohan/Downloads/16-0"

predictions_csv = os.path.join(
    base_dir, "16-06-30", "video_preds",
    "0532_CLAHE_with_pink_points.csv"
)
input_video_path = os.path.join(
    base_dir, "16-06-30", "video_preds", "labeled_videos",
    "0532_CLAHE_with_pink_points_labeled.mp4"
)
output_video_path = os.path.join(
    base_dir, "16-06-30", "video_preds", "labeled_videos",
    "0532_CLAHE_with_pink_points_labeled_with_purple_customfit.mp4"
)

# ─────────────────────────────────────────────
# PARAMETERS
# ─────────────────────────────────────────────
num_rays = 200          # number of interpolated purple points per frame
y_min = 386             # keep only keypoints with y in [y_min, y_max)
y_max = 664

# ─────────────────────────────────────────────
# 1. LOAD PREDICTIONS CSV
# ─────────────────────────────────────────────
df = pd.read_csv(predictions_csv, header=[0, 1, 2])

scorer = df.columns[1][0]

bodypart_names = []
for col in df.columns[1:]:
    if col[2] == "x":
        bodypart_names.append(col[1])

x_cols = [(scorer, bp, "x") for bp in bodypart_names]
y_cols = [(scorer, bp, "y") for bp in bodypart_names]

print(f"Loaded {len(df)} frames, {len(bodypart_names)} bodyparts")

# ─────────────────────────────────────────────
# 2. OPEN VIDEO
# ─────────────────────────────────────────────
cap = cv2.VideoCapture(input_video_path)
fps = cap.get(cv2.CAP_PROP_FPS)
w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
print(f"Video: {w}x{h} @ {fps} fps, {total_frames} frames")

# ─────────────────────────────────────────────
# 3. PER-FRAME: custom_function fit → purple points
# ─────────────────────────────────────────────
purple_points_over_time = []
cached_popt = None

# curve_fit bounds (from your original code)
lower_param_bounds = [-np.inf, -np.inf, -np.inf, -np.inf,
                       42.0, 0.0, 0.0, 0, -np.pi, 0, -np.pi, 42.0]
upper_param_bounds = [np.inf, np.inf, np.inf, np.inf,
                      np.inf, 0.0025, np.inf, 12.0, np.pi, 12.0, np.pi, np.inf]
param_bounds = (lower_param_bounds, upper_param_bounds)

loop_start = time.perf_counter()
frame_idx = 0

while cap.isOpened():
    ret, frame = cap.read()
    if not ret or frame_idx >= len(df):
        break

    if frame_idx % 50 == 0:
        print(f"  custom fit: {frame_idx} / {total_frames}")

    # Extract x, y for this frame and sort by x
    xs_raw = np.array([float(df.iloc[frame_idx][xc]) for xc in x_cols])
    ys_raw = np.array([float(df.iloc[frame_idx][yc]) for yc in y_cols])

    sort_idx = np.argsort(xs_raw)
    xs_raw = xs_raw[sort_idx]
    ys_raw = ys_raw[sort_idx]

    # Filter by y range and x within frame
    mask = (xs_raw >= 0) & (xs_raw < w) & (ys_raw >= y_min) & (ys_raw < y_max)
    true_x = xs_raw[mask]
    true_y = ys_raw[mask]

    if len(true_x) < 12:
        # Not enough points — emit NaNs
        purple_points_over_time.append([np.nan] * (2 * num_rays))
        frame_idx += 1
        continue

    # Shift x so the fit starts at 0 (matches your original code)
    x_shifted = true_x - np.min(true_x)
    the_mean = np.mean(true_y)

    initial_guess = [0.0, 1.0, 0.0, the_mean,
                     85.0, 0.00125, 0.0,
                     0.02, 0.0, 0.02, 0.0, 85.0]

    try:
        popt, _, _, mesg, _ = curve_fit(
            custom_function, x_shifted, true_y,
            p0=initial_guess, bounds=param_bounds, full_output=True
        )
        cached_popt = popt
    except Exception:
        print(f"  frame {frame_idx}: curve_fit failed, using cached params")
        if cached_popt is None:
            purple_points_over_time.append([np.nan] * (2 * num_rays))
            frame_idx += 1
            continue
        popt = cached_popt

    # Evaluate on smooth grid
    x_smooth = np.linspace(np.min(true_x), np.max(true_x), num_rays)
    x_smooth_shifted = x_smooth - np.min(true_x)
    y_smooth = custom_function(x_smooth_shifted, *popt)

    # Store interleaved [x1, y1, x2, y2, ...]
    purple_row = []
    for i in range(num_rays):
        purple_row.append(float(x_smooth[i]))
        purple_row.append(float(y_smooth[i]))
    purple_points_over_time.append(purple_row)

    frame_idx += 1

cap.release()
purple_points_over_time = np.array(purple_points_over_time, dtype=float)
print(f"Custom fitting done ({frame_idx} frames in {time.perf_counter()-loop_start:.1f}s)")

# ─────────────────────────────────────────────
# 4. DRAW PURPLE POINTS ONTO A COPY
# ─────────────────────────────────────────────
def draw_points_on_video(
    src_path, dst_path, points,
    radius=4, color=(255, 0, 255), thickness=-1
):
    """
    Read src_path, overlay points on each frame, write to dst_path.
    points: (N_frames, 2*N_pts) — interleaved [x1,y1,x2,y2,...].
    """
    cap = cv2.VideoCapture(src_path)
    fps = cap.get(cv2.CAP_PROP_FPS)
    w_v = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    h_v = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fourcc = cv2.VideoWriter_fourcc(*"mp4v")
    out = cv2.VideoWriter(dst_path, fourcc, fps, (w_v, h_v))

    i = 0
    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break
        if i % 50 == 0:
            print(f"  drawing: {i}")
        if i < len(points):
            coords = points[i].reshape(-1, 2)
            for x, y in coords:
                if not np.isnan(x) and not np.isnan(y):
                    cv2.circle(frame, (int(x), int(y)), radius, color, thickness)
        out.write(frame)
        i += 1

    cap.release()
    out.release()
    print(f"Saved annotated video to: {dst_path}")

draw_points_on_video(input_video_path, output_video_path, purple_points_over_time)
print("All done!")
```
